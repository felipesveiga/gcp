<h2 style='font-size:30px'> The GCP Resource Hierarchy: Organizing Your Digital Assets</h2>
<div> 
    <ul style='font-size:20px'> 
        <li>
            O Google Cloud provê uma organização hierárquica sobre seus recursos.
        </li>
        <li>
            A raiz seria o nível <i> Organization</i>, que contempla a empresa como um todo. Abaixo dela, temos <i> Folders</i>, e dentro dessas os <i> Projects</i>.
        </li>
        <li>
            Por fim, dentro dos projetos, estarão alocados os recursos do GCP de fato (VM's, DB's, etc).
        </li>
    </ul>
</div>

<h3 style='font-size:30px;font-style:italic'> 
    Policy Inheritance
</h3>
<div> 
    <ul style='font-size:20px'>
        <li>
            Importante ressaltar que, ao adentrarmos na parte de permissionamento, uma vez aplicada uma política num nível hierárquico, essa será valida para todas as entidades em níveis menores.
        </li>
        <li>
            Por exemplo, uma permissão a nível org será válida para todos os projects.
        </li>
     </ul>
</div>

<h2 style='font-size:30px'> Controlling Your Environment with Organization Policies</h2>
<div> 
    <ul style='font-size:20px'> 
        <li>
            No Google Cloud, existem duas entidades de definição de ações que devem ser diferenciadas.
            <ul> 
                <li>
                    <i> Organization Policy:</i> Definição <strong> do que</strong> pode ou não ser feito, dentro da organização.
                </li>
                <li>
                    <i> IAM</i>: Definição <strong> de quem</strong> pode fazer certas ações, dentro de um dado escopo (org, projeto, etc).
                </li>
            </ul>
        </li>
    </ul>
</div>
