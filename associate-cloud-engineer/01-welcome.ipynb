<h2 style='font-size:30px'> The Big Picture: GCP's Global Infrastructure (Regions, Zones, Edge Locations)</h2>
<div> 
    <ul style='font-size:20px'> 
        <li>
            Apresentação de conceitos de hierarquia dos pontos de presença do Google Cloud.
        </li>
        <li>
            De maneira geral, são bem parecidos com a AWS (Regions, Zones, Edge Locations).
        </li>
        <li>
            Edge locations seriam pontos de presença da rede interna do GCP voltados a manter proximidade com os consumidores das nossas aplicações, reduzindo a latência do seu uso. 
        </li>
    </ul>
</div>

<h2 style='font-size:30px'> A Guided Tour of the Google Cloud Console</h2>
<div> 
    <ul style='font-size:20px'>
