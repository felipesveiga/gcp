<h2 style='font-size:30px'> The GCP Resource Hierarchy: Organizing Your Digital Assets</h2>
<div> 
    <ul style='font-size:20px'> 
        <li>
            O Google Cloud provê uma organização hierárquica sobre seus recursos.
        </li>
        <li>
            A raiz seria o nível <i> Organization</i>, que contempla a empresa como um todo. Abaixo dela, temos <i> Folders</i>, e dentro dessas os <i> Projects</i>.
        </li>
        <li>
            Por fim, dentro dos projetos, estarão alocados os recursos do GCP de fato (VM's, DB's, etc).
        </li>
    </ul>
</div>
